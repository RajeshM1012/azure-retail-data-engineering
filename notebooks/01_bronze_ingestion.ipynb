# 01 - Bronze Ingestion
Read raw retail CSV files and persist them as Delta tables.

```python
from pyspark.sql import functions as F
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('RetailBronzeIngestion').getOrCreate()

base_path = '/mnt/retail'  # Replace with ADLS Gen2 path in Azure
sales_path = f'{base_path}/input/sales.csv'
customers_path = f'{base_path}/input/customers.csv'
products_path = f'{base_path}/input/products.csv'
```

```python
sales_df = (spark.read.option('header', True).option('inferSchema', True).csv(sales_path))
customers_df = (spark.read.option('header', True).option('inferSchema', True).csv(customers_path))
products_df = (spark.read.option('header', True).option('inferSchema', True).csv(products_path))

sales_df = sales_df.withColumn('ingestion_timestamp', F.current_timestamp())
customers_df = customers_df.withColumn('ingestion_timestamp', F.current_timestamp())
products_df = products_df.withColumn('ingestion_timestamp', F.current_timestamp())
```

```python
sales_df.write.format('delta').mode('append').save(f'{base_path}/bronze/sales')
customers_df.write.format('delta').mode('append').save(f'{base_path}/bronze/customers')
products_df.write.format('delta').mode('append').save(f'{base_path}/bronze/products')
```
